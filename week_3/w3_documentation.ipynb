## Scenario

A student has marks for three quizzes. The program should:

1. Ask the user to enter three quiz marks.
2. Calculate the average mark.
3. Display the average.
4. Decide whether the student passes or fails.
5. Let the user enter another student's marks.

---
## 1. Identify the Components

### 1.1 Inputs

| Input | Type | Description |
|---|---|---|
| `quiz_1` | float | Mark for Quiz 1 |
| `quiz_2` | float | Mark for Quiz 2 |
| `quiz_3` | float | Mark for Quiz 3 |
| `choice` | string | `"y"` to enter another student, anything else to stop |

### 1.2 Process

1. Add the three marks together.
2. Divide the total by 3 to get the average:  `average = (quiz_1 + quiz_2 + quiz_3) / 3`
3. Compare the average with the pass mark of 50:
   - `average >= 50` means **PASS**
   - `average < 50` means **FAIL**
4. Ask whether to continue, and repeat while the answer is `"y"`.

### 1.3 Output

- The student's **average mark** (2 decimal places)
- The result: **PASS** or **FAIL**
- `"Program Ended"` once the user stops

---
## 2. Design the Algorithm

### 2.1 Pseudocode

```
BEGIN
    SET choice = "y"

    WHILE choice == "y" DO
        INPUT quiz_1
        INPUT quiz_2
        INPUT quiz_3

        average = (quiz_1 + quiz_2 + quiz_3) / 3
        OUTPUT average

        IF average >= 50 THEN
            OUTPUT "PASS"
        ELSE
            OUTPUT "FAIL"
        ENDIF

        INPUT choice        // "Continue? Select Y/N"
    ENDWHILE

    OUTPUT "Program Ended"
END
```

### 2.2 Algorithm Structure

All three basic control structures appear in this program.

| Structure | What it means | Where it appears in the program |
|---|---|---|
| **2.2.1 Sequence** | Steps run one after another, top to bottom | Read `quiz_1`, then `quiz_2`, then `quiz_3`, then calculate `average`, then print it |
| **2.2.2 Selection** | The program picks a path based on a condition | `if average >= 50:` prints PASS, `else:` prints FAIL |
| **2.2.3 Iteration** | A block of code repeats while a condition holds | `while choice == "y":` repeats the whole process for each student |

Think of it as nesting: the **iteration** loop wraps everything. Inside it the steps run in **sequence**, and the sequence includes one **selection** step that decides PASS or FAIL.

## 3. The Python Program (`Student_Ave_Calc.py`)

This is the completed template. It uses `input()`, so running it here asks you to type marks.

```python
# Student_Ave_Calc.py
# Tutorial 2 (Week 3) - Student Average Calculator
# Name: Nasir Mohamed Abdelgafar Elyas
# Student ID: 202405010110


# Input validation helper: keeps asking until the mark is a number from 0 to 100.
# Without it, typing "abc" or pressing Enter crashes the program with a ValueError,
# and impossible marks like 150 or -20 would be used in the average.
def get_mark(prompt):
    while True:
        try:
            mark = float(input(prompt))
        except ValueError:
            print("Please enter a number.")
            continue
        if 0 <= mark <= 100:   # also rejects "nan", because nan fails every comparison
            return mark
        print("Mark must be between 0 and 100.")


choice = "y"

# Iteration: keep repeating the whole process for each student
# as long as the user's choice is "y"
while choice == "y":

    # Sequence (input): ask the user for the three quiz marks.
    # get_mark() converts each one to a number with float() and checks it is 0-100
    quiz_1 = get_mark("Enter Quiz 1 mark: ")
    quiz_2 = get_mark("Enter Quiz 2 mark: ")
    quiz_3 = get_mark("Enter Quiz 3 mark: ")

    # Sequence (process + output): add the three marks, divide by 3
    # to get the average, then display it to 2 decimal places
    average = (quiz_1 + quiz_2 + quiz_3) / 3
    print(f"Average mark: {average:.2f}")

    # Selection: an average of 50 or more is a pass, anything below 50 is a fail
    if average >= 50:
        print("Result: PASS")
    else:
        print("Result: FAIL")

    # Ask whether to enter another student's marks.
    # .strip().lower()[:1] keeps only the first letter, so "Y", "y", "yes" and "Yes"
    # all become "y" and the loop continues
    choice = input("Continue? Select Y/N: ").strip().lower()[:1]

print("Program Ended")
```

**Why `get_mark()`?** `input()` always returns text, and `float()` turns it into a number, but `float("abc")` crashes the program. `get_mark()` catches that, asks again, and rejects marks outside 0–100.

**Why `.strip().lower()[:1]`?** The loop checks for a lowercase `"y"`. Keeping only the first letter means `Y`, `y`, `yes` and `Yes` all continue the program.

## 4. Test with Sample Inputs

The cell below runs the same calculation on fixed sample marks, so the results show without typing anything. It includes the boundary case where the average is exactly 50.

````python
def student_result(quiz_1, quiz_2, quiz_3):
    average = (quiz_1 + quiz_2 + quiz_3) / 3
    result = "PASS" if average >= 50 else "FAIL"
    return average, result

samples = [
    (80, 70, 60),    # clear pass
    (40, 45, 50),    # clear fail
    (50, 50, 50),    # boundary: exactly 50 -> PASS
    (49, 50, 50),    # just below 50 -> FAIL
    (72.5, 88, 91),  # decimal marks
    (100, 100, 100), # maximum marks
    (0, 0, 0),       # minimum marks
]

print(f"{'Quiz 1':>6} {'Quiz 2':>6} {'Quiz 3':>6} | {'Average':>7} | Result")
print("-" * 42)
for q1, q2, q3 in samples:
    avg, res = student_result(q1, q2, q3)
    print(f"{q1:>6} {q2:>6} {q3:>6} | {avg:>7.2f} | {res}")
````
